```python
from pathlib import Path

import gym_pusht  # noqa: F401
import gymnasium as gym
import imageio
import numpy as np
import torch

# from lerobot.common.policies.diffusion.modeling_diffusion import DiffusionPolicy
from lerobot.common.policies.act.modeling_act import ACTPolicy

from glob import glob
from PIL import Image
import torch
from torchvision.transforms import ToTensor
import tqdm
```

```python
# Select your device
device = "cuda"

# Provide the [hugging face repo id](https://huggingface.co/lerobot/diffusion_pusht):
pretrained_policy_path = "outputs/train/act_retraction_predict_js_30hz_eff_vel_depimg_pc/checkpoints/040000/pretrained_model"
# OR a path to a local outputs/train folder.
# pretrained_policy_path = Path("outputs/train/example_pusht_diffusion")
policy = ACTPolicy.from_pretrained(pretrained_policy_path, local_files_only=True)
# Reset the policy and environments to prepare for rollout
policy.reset()
```

```python
state = torch.zeros(19, device=device).unsqueeze(0)
left_img = torch.zeros(3, 256, 256, device=device).unsqueeze(0)
right_img = torch.zeros(3, 256, 256, device=device).unsqueeze(0)
dissection_target = torch.zeros(8, device=device).unsqueeze(0)
point_cloud = torch.rand(4096, 3, device=device).unsqueeze(0)

fake_observation = {
        "observation.state": state,
        "observation.images.cam_left": left_img,
        "observation.images.cam_right": right_img,
        "observation.dissection_tar": dissection_target,
        "observation.point_cloud": point_cloud,
    }
```

```python
 # Predict the next action with respect to the current observation
with torch.inference_mode():
    action = policy.select_action(fake_observation)
# Print the predicted action
print("Predicted Action:", action)
```
